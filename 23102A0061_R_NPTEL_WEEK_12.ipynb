**Subdivided or Component Bar Diagram**


```r
cust <- matrix(
  c(2, 20, 30,
    26, 53, 40,
    42, 15, 25,
    30, 75, 100),
  nrow = 4,
  byrow = TRUE
)

colnames(cust) <- c("Shop 1", "Shop 2", "Shop 3")
rownames(cust) <- c("Day 1", "Day 2", "Day 3", "Day 4")

print(cust)

barplot(
  cust,
  main = "Customers Visiting Three Shops",
  xlab = "Shops",
  ylab = "Number of Customers",
  col = c("red", "green", "orange", "brown"),
  legend.text = rownames(cust),
  args.legend = list(x = "topleft", cex = 0.7)
)
```

**Pie Diagram: Gender Distribution**


```r
gender <- c(1, 2, 1, 2, 1, 1, 1, 2, 1, 1)

gender_count <- table(
  factor(gender, levels = c(1, 2),
         labels = c("Male", "Female"))
)

print(gender_count)

pie(
  gender_count,
  labels = paste0(
    names(gender_count), ": ", gender_count
  ),
  col = c("skyblue", "pink"),
  main = "Gender Distribution"
)
```

**Example 1: Function Using Two Data Vectors**


```r
example1 <- function(x, y) {
  if (length(x) != length(y)) {
    stop("x and y must have the same length")
  }

  if (any(y == 0)) {
    stop("Values of y must not be zero")
  }

  g <- sum(x^2) / sum(y^2)
  h <- sum((x / y)^2)

  cat("The value of g is:", g, "\n")
  cat("The value of h is:", h, "\n")

  return(c(g = g, h = h))
}

# Test the function
x <- c(10, 20, 30)
y <- c(1, 2, 3)

example1(x, y)
```

**Example 2: Defining and Calling Functions**


```r
g <- function(x, y) {
  if (y <= 0) {
    stop("y must be positive")
  }

  (x + log(y)) / y
}

f <- function(x, y) {
  z <- g(x, y)

  ((z^2) / (5 + z^3)) * exp(z)^(2/3)
}

# Test with values from the notes
print(f(10, 20))
print(f(1896, 23454))
```

**Example 3: Piecewise Function and Plot**


```r
f <- function(x) {
  if (x > 0) {
    exp((x + log(1 + x^3)) / x^2)
  } else if (x == 0) {
    10
  } else {
    (2 + x^3) / x
  }
}

# Test individual values
print(f(123))
print(f(-123))
print(f(0))
print(f(8))
print(f(-4))

# Generate values and plot the function
x_values <- seq(-1, 5, by = 0.2)

y_values <- sapply(x_values, f)

plot(
  x_values, y_values,
  type = "l",
  main = "Plot of Piecewise Function",
  xlab = "x",
  ylab = "f(x)",
  col = "blue",
  lwd = 2
)
```
