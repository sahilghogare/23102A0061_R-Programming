**Factors**

```r
x <- factor(
  c(
    "lemonade",
    "lemonade",
    "juice",
    "lemonade",
    "water"
  )
)
print(x)
```

```r
print(levels(x))
```

```r
x <- c(
  3, 4, 5, 6,
  1, 2, 3, 3,
  4, 4, 5, 6
)

y <- as.factor(x)
print(y)
```

```r
print(levels(y))
```

```r
y <- c(
  1, 4, 3, 5,
  4, 2, 4
)

print(y)
```

```r
possible.dieface <- c(
  1, 2, 3, 4, 5, 6
)

print(possible.dieface)
```

```r
labels.dieface <- c(
  "one",
  "two",
  "three",
  "four",
  "five",
  "six"
)
print(labels.dieface)
```

```r
facy <- factor(
  y,
  levels = possible.dieface,
  labels = labels.dieface
)

print(facy)
```

```r
print(levels(facy))
```

**Factors – Class and Unclass**

```r
print(class(9))

print(class("9"))

print(class(print))
```

```r
x <- matrix(
  nrow = 2,
  ncol = 2,
  data = 1:4
)

print(x)

print(class(x))
```

```r
brands <- c(
  "A",
  "A",
  "B",
  "B",
  "B",
  "B",
  "C"
)

brands_fac <- factor(brands)

print(brands_fac)
```

```r
print(unclass(brands_fac))
```

```r
colours <- c(
  "blue",
  "green",
  "red"
)

print(colours)
```

```r
print(colours[unclass(brands_fac)])
```

```r
x <- factor(
  c(
    "lemonade",
    "lemonade",
    "juice",
    "lemonade",
    "water"
  ),
  levels = c(
    "water",
    "juice",
    "lemonade"
  )
)

print(x)
```

```r
print(levels(x))
```

```r
print(unclass(x))
```

```r
income <- ordered(
  c(
    "high",
    "high",
    "low",
    "medium",
    "medium"
  ),
  levels = c(
    "low",
    "medium",
    "high"
  )
)

print(income)
```

```r
print(levels(income))
```

```r
print(unclass(income))
```
